```python
%%writefile app.py

import streamlit as st
import pickle
import numpy as np

# -----------------------------
# Load trained model
# -----------------------------
with open("brain.pkl", "rb") as f:
    model = pickle.load(f)

# -----------------------------
# Encoding maps (same as training)
# -----------------------------
gender_map = {"female": 0, "male": 1, "others": 2}

smoking_map = {
    "No Info": 0,
    "current": 1,
    "ever": 2,
    "former": 3,
    "never": 4,
    "not current": 5
}

# -----------------------------
# App Title
# -----------------------------
st.title("🧬 Diabetes Prediction System")
st.write("Enter patient details to predict Diabetes")

# -----------------------------
# User Inputs
# -----------------------------
age = st.number_input("Age", min_value=1, max_value=120, value=30)

gender = st.selectbox("Gender", ["female", "male", "others"])

hypertension = st.selectbox("Hypertension", [0, 1])
heart_disease = st.selectbox("Heart Disease", [0, 1])

smoking = st.selectbox("Smoking History",
                       ["No Info", "current", "ever", "former", "never", "not current"])

bmi = st.number_input("BMI", min_value=10.0, max_value=60.0, value=25.0)

hba1c = st.number_input("HbA1c Level", min_value=3.0, max_value=15.0, value=5.5)

glucose = st.number_input("Blood Glucose Level", min_value=50, max_value=300, value=100)

# -----------------------------
# Encode inputs
# -----------------------------
gender_encoded = gender_map[gender]
smoking_encoded = smoking_map[smoking]

input_data = np.array([
    age,
    gender_encoded,
    hypertension,
    heart_disease,
    smoking_encoded,
    bmi,
    hba1c,
    glucose
]).reshape(1, -1)

# -----------------------------
# Prediction
# -----------------------------
if st.button("Predict Diabetes"):
    prediction = model.predict(input_data)[0]

    if prediction == 1:
        st.error("⚠️ High Risk: Patient is likely Diabetic")
    else:
        st.success("✅ Low Risk: Patient is NOT Diabetic")
```
